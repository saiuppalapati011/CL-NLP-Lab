# 4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.

## 4.1.1 Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

```python
import re
import spacy
import pandas as pd
from collections import Counter

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# Read the input file
with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# TERM FREQUENCY ANALYSIS


# Process the text using spaCy
doc = nlp(text)

# Extract only alphabetic words and convert to lowercase
words = [token.text.lower() for token in doc if token.is_alpha]

# Calculate word frequencies
frequency = Counter(words)

# Save frequencies to CSV
df = pd.DataFrame(frequency.items(), columns=["Term", "Frequency"])
df.to_csv("term_frequency.csv", index=False)

# Print all term frequencies
print("===== TERM FREQUENCIES =====\n")
print(df)


# NAMED ENTITY RECOGNITION

print("\n===== NAMED ENTITIES =====\n")

print("{:<35}{}".format("Entity", "Label"))
print("-" * 50)

for entity in doc.ents:
    print(f"{entity.text:<35}{entity.label_}")
```

## 4.1.2 Identify the top 10 most frequent terms and display their corresponding frequencies.

```python
print("\nTop 10 Most Frequent Terms\n")

# Get the top 10 most common words
top_10 = frequency.most_common(10)

# Display them
print("{:<20}{}".format("Term", "Frequency"))
print("-" * 30)

for term, count in top_10:
    print("{:<20}{}".format(term, count))
```

# 4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

## 4.2.1 Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

```python
import re
import csv

# Read the input file
with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Convert to lowercase
text = text.lower()

# Remove punctuation
text = re.sub(r"[^\w\s]", "", text)

# Tokenize
words = text.split()

# Calculate word frequency using a dictionary
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Save the frequencies to a CSV file
with open("term_frequency.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    # Header
    writer.writerow(["Term", "Frequency"])

    # Write each term and its frequency
    for term, count in frequency.items():
        writer.writerow([term, count])

print("Term frequencies saved to 'term_frequency.csv'")
```

## 4.2.2 Identify the top 10 most frequent terms and display their corresponding frequencies.

```python
# Sort the dictionary by frequency (highest to lowest)
sorted_frequency = sorted(
    frequency.items(),
    key=lambda item: item[1],
    reverse=True
)

# Display the top 10 most frequent terms
print("\nTop 10 Most Frequent Terms\n")
print("{:<20}{}".format("Term", "Frequency"))
print("-" * 30)

for term, count in sorted_frequency[:10]:
    print("{:<20}{}".format(term, count))
```

# 4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

```python
import math


# STEP 1: Read Multiple Documents

documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]


# STEP 2: Convert Text to Lowercase
# STEP 3: Remove Punctuation
# STEP 4: Tokenize the Documents

processed_docs = []

for doc in documents:

    # Convert to lowercase
    doc = doc.lower()

    # Remove punctuation using Regular Expression
    doc = re.sub(r"[^\w\s]", "", doc)

    # Tokenize the document using split()
    tokens = doc.split()

    # Store the tokenized document
    processed_docs.append(tokens)

# Display tokenized documents
print("========== TOKENIZED DOCUMENTS ==========\n")

for i, tokens in enumerate(processed_docs, start=1):
    print(f"Document {i}:")
    print(tokens)
    print()


# STEP 5: Calculate TF (Term Frequency)

tf_list = []

for tokens in processed_docs:

    # Dictionary to store word counts
    term_frequency = {}

    # Count each word
    for word in tokens:
        if word in term_frequency:
            term_frequency[word] += 1
        else:
            term_frequency[word] = 1

    # Total words in this document
    total_words = len(tokens)

    # Convert counts into TF values
    for word in term_frequency:
        term_frequency[word] = term_frequency[word] / total_words

    tf_list.append(term_frequency)

# Display TF
print("\n========== TERM FREQUENCY (TF) ==========\n")

for i, tf in enumerate(tf_list, start=1):
    print(f"Document {i}")

    print("{:<20}{}".format("Term", "TF"))

    for term, value in tf.items():
        print("{:<20}{:.3f}".format(term, value))

    print()


# STEP 6: Calculate DF (Document Frequency)

df = {}

for tokens in processed_docs:

    # Use a set so the same word is counted only once
    unique_words = set(tokens)

    for word in unique_words:

        if word in df:
            df[word] += 1
        else:
            df[word] = 1

print("\n========== DOCUMENT FREQUENCY (DF) ==========\n")

print("{:<20}{}".format("Term", "DF"))

for term, value in df.items():
    print("{:<20}{}".format(term, value))


# STEP 7: Calculate IDF (Inverse Document Frequency)

idf = {}

total_documents = len(processed_docs)

for word in df:

    idf[word] = math.log(total_documents / df[word])

print("\n========== INVERSE DOCUMENT FREQUENCY (IDF) ==========\n")

print("{:<20}{}".format("Term", "IDF"))

for term, value in idf.items():
    print("{:<20}{:.3f}".format(term, value))


# STEP 8:
# Calculate TF-IDF

print("\n========== TF-IDF ==========\n")

for i in range(total_documents):

    print(f"Document {i+1}\n")

    tfidf = {}

    # Calculate TF-IDF for every term
    for word in tf_list[i]:

        tfidf[word] = tf_list[i][word] * idf[word]

 # STEP 9:
 # Display Term, TF, DF, IDF and TF-IDF

    print("{:<18}{:<10}{:<8}{:<10}{:<10}".format(
        "Term",
        "TF",
        "DF",
        "IDF",
        "TF-IDF"
    ))

    print("-" * 60)

    for word in tfidf:

        print("{:<18}{:<10.3f}{:<8}{:<10.3f}{:<10.3f}".format(
            word,
            tf_list[i][word],
            df[word],
            idf[word],
            tfidf[word]
        ))

    # STEP 10: Identify Top 10 Highest TF-IDF Terms


    print("\nTop 10 Highest TF-IDF Terms\n")

    # Sort in descending order of TF-IDF score
    top_terms = sorted(
        tfidf.items(),
        key=lambda item: item[1],
        reverse=True
    )

    # Print top 10 terms
    for term, score in top_terms[:10]:
        print("{:<20}{:.3f}".format(term, score))

    print("\n" + "=" * 70 + "\n")
```
