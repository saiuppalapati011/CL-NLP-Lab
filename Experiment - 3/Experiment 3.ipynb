## Experiment 3.1:  Write a Python program to perform  Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.

`playing, played, plays, studies, studying, connected, connection, computers`

```python
from nltk.stem import PorterStemmer

# Create a Porter Stemmer object
stemmer = PorterStemmer()

# Given words
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connected",
    "connection",
    "computers"
]

print("{:<15}{}".format("Original Word", "Stem"))

for word in words:
    print("{:<15}{}".format(word, stemmer.stem(word)))
```

## Experiment 3.2:  Write a Python program to perform lemmatization on a given input words:

`cats,	dogs,   running,  runs,	ran, studies, studying,	better, children, mice,	went, ate,	
leaves,	caring,`

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "cats",
    "dogs",
    "running",
    "runs",
    "ran",
    "studies",
    "studying",
    "better",
    "children",
    "mice",
    "went",
    "ate",
    "leaves",
    "caring"
]

print("{:<15}{}".format("Original Word", "Lemma"))

for word in words:
    print("{:<15}{}".format(word, lemmatizer.lemmatize(word)))
```

## Experiment 3.2: Write a Python program using Regular Expressions (Regex) to extract and process the following information from a text:

- Email addresses
- URLs
- Mobile numbers
- Hashtags
- Mentions

Input Text:

Welcome to the Natural Language Processing workshop! For more information, visit https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop and @PythonLearner. Share your experience using #NLP, #Python, and #MachineLearning. You can also visit https://github.com/NLPWorkshop for the latest updates.

```python
import re

text = """
Welcome to the Natural Language Processing workshop! For more information, visit
https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at
nlpworkshop@gmail.com or support@python.org. For registration, call
+91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop and
@PythonLearner. Share your experience using #NLP, #Python, and
#MachineLearning. You can also visit https://github.com/NLPWorkshop for the
latest updates.
"""

# Email Addresses
emails = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text)

# URLs
urls = re.findall(r"https?://\S+|www\.\S+", text)

# Mobile Numbers
mobile_numbers = re.findall(r"\+91-\d{10}|\b\d{10}\b", text)

# Hashtags
hashtags = re.findall(r"#\w+", text)

# Mentions
mentions = re.findall(r"@\w+", text)

print("Email Addresses")
print(emails)

print("\nURLs")
print(urls)

print("\nMobile Numbers")
print(mobile_numbers)

print("\nHashtags")
print(hashtags)

print("\nMentions")
print(mentions)
```
