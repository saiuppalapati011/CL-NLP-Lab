# Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:

## Convert text to lowercase. (Also display how many uppercase?)

```python
import re
import string
```

```python
with open("2.1_text_data.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
print("---Original text-----")
print(text)
```

```python
uppercase_count = sum(1 for ch in text if ch.isupper())
text = text.lower()
print("Count of number of upper case letters:", uppercase_count)
print("---Lowercase Text----")
print(text)
```

## Remove punctuation.(Also display how many different types of punctuations?)

```python
punctuation_found = [ch for ch in text if ch in string.punctuation]
unique_punctuation = set(punctuation_found)
text = text.translate(str.maketrans('', '', string.punctuation))
print(len(unique_punctuation))
print(text)
```

## Remove numbers.(Also display how many numbers?)

```python
number_count = sum(1 for ch in text if ch.isdigit())
text = re.sub(r'\d+', '', text)
print(number_count)
print(text)
```

## Remove extra whitespace. (How much extra whitespace is present?)

```python
text = re.sub(r'\s+', ' ', text).strip()
```

## Display the cleaned text.    

```python
print("----- Final Cleaned test ------")
print(text)
```

# Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

```python
 # Tokenization using NLTK
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Read the text file
with open("2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Sentence Tokenization
sentences = sent_tokenize(text)

# Word Tokenization
words = word_tokenize(text)

print("---- SENTENCES ----")
for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")

print("\n---- WORDS ----")
print(words)
```

```python
# tokenization using Spacy

import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

# Read the text file
with open("2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

doc = nlp(text)

print("---- SENTENCES ----")
for i, sent in enumerate(doc.sents, start=1):
    print(f"{i}. {sent.text}")

print("\n---- WORDS ----")
words = [token.text for token in doc]

print(words)
```

```python
import re
# Read the text file
with open("2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

sentences = re.split(r'[.!?]+', text)

# Remove empty strings
sentences = [s.strip() for s in sentences if s.strip()]

words = re.findall(r"\b[\w'-]+\b", text)

print("---- SENTENCES ----")
for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")

print("\n---- WORDS ----")
print(words)
```

## Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
nlp = spacy.load("en_core_web_sm")

# Read file
with open("2.3_clean_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

doc = nlp(text)

tokens = [token.text for token in doc]

stop_words = [token.text for token in doc if token.is_stop]

filtered_tokens = [token.text for token in doc if not token.is_stop]

print("---- ORIGINAL TOKENS ----")
print(tokens)

print("\n---- STOP WORDS ----")
print(stop_words)

print("\n---- FILTERED TOKENS ----")
print(filtered_tokens)
```
